# DNA Sequence Analyzer

## Project objective
To analyse a DNA sequence using Python and calculate
its length, nucleotide composition, GC content,
and reverse complement.

## Tools
- Python
- Google Colab
- Matplotlib

```python

# DNA Sequence Analyzer
# MSc Bioinformatics Mini Project

dna_sequence = "ATGCGATCGATCGGCTAAGC"

print("DNA Sequence:")
print(dna_sequence)

print("\nSequence Length:")
print(len(dna_sequence))
```

```python
# GC content calculation

# Step 3: Nucleotide Counting

dna_sequence = "ATGCGATCGATCGGCTAAGC"

A_count = dna_sequence.count("A")
T_count = dna_sequence.count("T")
G_count = dna_sequence.count("G")
C_count = dna_sequence.count("C")

print("Nucleotide Counts")
print("A:", A_count)
print("T:", T_count)
print("G:", G_count)
print("C:", C_count)

# Calculate GC content

gc_content = ((G_count + C_count) / len(dna_sequence)) * 100

print("\nGC Content:", round(gc_content, 2), "%")

# Calculate AT content

at_content = ((A_count + T_count) / len(dna_sequence)) * 100

print("AT Content:", round(at_content, 2), "%")
```

```python
# reverse complement analysis

# Step 4: Reverse Complement Analysis

complement = ""

for base in dna_sequence:
    if base == "A":
        complement += "T"
    elif base == "T":
        complement += "A"
    elif base == "G":
        complement += "C"
    elif base == "C":
        complement += "G"

reverse_complement = complement[::-1]

print("Original DNA:", dna_sequence)
print("Complement:", complement)
print("Reverse Complement:", reverse_complement)
```

```python
#Start and Stop Codon Identification

# Step 5: Start and Stop Codon Identification

start_codon = "ATG"
stop_codons = ["TAA", "TAG", "TGA"]

print("Start Codon:", start_codon)

print("\nStop Codons Found:")

for i in range(len(dna_sequence) - 2):
    codon = dna_sequence[i:i+3]

    if codon in stop_codons:
        print(codon, "at position", i+1)

# Check for start codon

if start_codon in dna_sequence:
    print("\nStart codon ATG is present.")
else:
    print("\nStart codon ATG is absent.")
```

```python
#DNA Nucleotide Frequency Graph

# Step 6: Nucleotide Frequency Visualization

import matplotlib.pyplot as plt

nucleotides = ["A", "T", "G", "C"]

counts = [
    dna_sequence.count("A"),
    dna_sequence.count("T"),
    dna_sequence.count("G"),
    dna_sequence.count("C")
]

plt.figure(figsize=(7, 5))

plt.bar(nucleotides, counts)

plt.title("DNA Nucleotide Frequency")
plt.xlabel("Nucleotide")
plt.ylabel("Frequency")

plt.show()

```

```python

# Step 7: Final DNA Analysis Report

print("=" * 40)
print("       DNA SEQUENCE ANALYSIS REPORT")
print("=" * 40)

print("\n1. Original DNA Sequence:")
print(dna_sequence)

print("\n2. Sequence Length:")
print(len(dna_sequence), "nucleotides")

print("\n3. Nucleotide Composition:")
print("Adenine (A):", A_count)
print("Thymine (T):", T_count)
print("Guanine (G):", G_count)
print("Cytosine (C):", C_count)

print("\n4. GC Content:")
print(round(gc_content, 2), "%")

print("\n5. AT Content:")
print(round(at_content, 2), "%")

print("\n6. Reverse Complement:")
print(reverse_complement)

print("\n7. Start Codon:")
print("ATG present:", "ATG" in dna_sequence)

print("\n8. Stop Codons:")
for codon in ["TAA", "TAG", "TGA"]:
    print(codon, ":", dna_sequence.count(codon))

print("\n" + "=" * 40)
print("          ANALYSIS COMPLETED")
print("=" * 40)
```
